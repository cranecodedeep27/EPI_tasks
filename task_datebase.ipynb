```python
import pandas as pd

# дано:
cats = [('Мартин', 5, 'Алексей', 'Егоров'),
    	('Фродо', 3, 'Анна', 'Самохина'),
    	('Вася', 4, 'Андрей', 'Белов'),
    	('Муся', 7, 'Игорь', 'Бероев'),
    	('Изольда', 2, 'Игорь', 'Бероев'),
    	('Снейп', 1, 'Марина', 'Апраксина'),
    	('Лютик', 4, 'Виталий', 'Соломин'),
    	('Снежок', 3, 'Марина', 'Апраксина'),
    	('Марта', 5, 'Сергей', 'Колесников'),
    	('Буся', 12, 'Алена', 'Федорова'),
    	('Джонни', 10, 'Игорь', 'Андропов'),
    	('Мурзик', 1,'Даниил', 'Невзоров'),
    	('Барсик', 2, 'Виталий', 'Соломин'),
    	('Рыжик', 7, 'Владимир', 'Медведев'),
        ('Матильда', 8, 'Андрей', 'Белов'),
    	('Гарфилд', 3, 'Александр', 'Березуев')]
```

Найти: 1)Самого молодого питомца
2)Самого старого питомца
3)Внести данные еще об одном питомце
4)Вывести на экран всех владельцев с определенным именем и клички их животных
5)Вычислить средний возраст питомцев

##### DataFrame - это двумерный набор данных (в виде таблички)

```python
columns = ['Кличка питомца', 'Возраст питомца', 'Имя владельца', 'Фамилия владельца']

data_base = pd.DataFrame(cats, columns=columns)
data_base.head()  # визуализация первых 5 значений
```

```python
# 1) Самый молодой питомец
age_pet = data_base["Возраст питомца"]
columns1 = ["Кличка питомца", "Возраст питомца"]

youngest_pet = data_base.loc[age_pet == age_pet.min(), columns1]
# атрибут .loc - это индексация по названию колонок

youngest_pet
```

```python
# 2) Самый старый питомец

oldest_pet = data_base.loc[age_pet == age_pet.max(), columns1]

oldest_pet
```

```python
# 3) Внести данные ещё об одном питомце
tuple_pet = ("Кузя", 4, "Дмитрий", "Cмирнов")

pet_df = pd.DataFrame([tuple_pet], columns=data_base.columns)

frames = [data_base, pet_df]

update_data_base = pd.concat(frames, ignore_index=True)
# .concate - это конкантенация табличек в пандасе

update_data_base
```

```python
# 5) вывести на экран владельцев с определённым именем и клички их питомцев
# пусть будет имя владельца: 'Даниил'

specific_name = update_data_base.loc[update_data_base["Имя владельца"] == "Даниил", ["Кличка питомца", "Имя владельца"]]

specific_name
```

```python
# вычислить средний возраст питомцев

update_data_base.describe()  
# функция .describe() позволяет посмотреть основные характеристики датафрейма (работает только с числовыми типами данных)
# Средний это строка mean
```
