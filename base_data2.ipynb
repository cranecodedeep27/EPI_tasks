### База - "климат" (15 записей)

```python
import pandas as pd
import numpy as np
from random import randint, uniform, choice
```

```python
n = 15 # количество записей
base_data_climate = {
    'код местности': np.random.randint(1, 4, size=n),
    'год наблюдения': np.random.randint(2020, 2026, size=n),
    'абсолютный максимум температуры': np.random.uniform(30, 41, size=n),
    'число дней без осадков': np.random.uniform(80, 141, size=n),
    'средняя температура за год': np.random.uniform(5, 16, size=n)
}

df = pd.DataFrame(base_data_climate)
df
```

```python
# 1) напечатать список кодов местностей с числом дней без осадков не менее 100 и средней температурой не более 8 градусов
a = df.loc[
    (df['число дней без осадков'] >= 100) & (df['средняя температура за год'] <= 8),
    'код местности'
]

a
```

```python
# 2) определить код местности с наибольшим абсолютным максимумом температуры
b = df.loc[df['абсолютный максимум температуры'].idxmax(), 'код местности']  # .idxmax() - возвращает индекс строки в которой находится максимальная температура
print(b)
```

### База "Группа" (12 записей)

```python
n = 12
fio = ["Захаров", "Петров", "Смирнов","Сидоров", "Смирнова", "Пупкин", "Иванов", "Чистова", "Цой", "Обаренов", "Ежиков", "Гусев"]
sex = ["муж", "жен", "муж", "муж", "жен", "муж", "муж", "жен", "муж", "муж","муж", "муж"]
base_data_group = {
    'номер по списку': np.arange(1, n+1),
    'Ф.И.О': fio,
    'год рождения': np.random.randint(1960, 2010, size=n),
    'пол': sex,
    'средний балл за сессию': np.random.uniform(2, 5, size=n),
    'наличие пересдач экзаменов': np.random.choice([True, False], size=n)
}
```

```python
# 2) изменить запись о студенте, 6-м по номеру в списке
df.loc[df['номер по списку'] == 6, 
       ['Ф.И.О', 'год рождения', 'пол', 'средний балл за сессию', 'наличие пересдач экзаменов']] = ['Конев', 2005, 'муж', 4.5, False]

df
```

### База - "команда" (9 записей)

```python
n = 9 
coaches = ["Жуков", "Хомяков", "Старостин", "Минаев", "Минин", "Пожарский", "Соловьев", "Гудини", "Потолков"]
teams = ["Барселона", "Реал", "Медвежонск", "Спартак", "Мадрид", "Звери", "Атлетико", "Питоновские", "The bests"]
data_base_team = {
    "Название команды": teams, 
    "Тренер": coaches, 
    "Место в прошлом сезоне": np.random.randint(1, 10, size=n),
    "Среднее число очков за игру": np.random.randint(30, 120, size=n), 
    "Средний возраст игроков": np.random.uniform(20, 40, size=n)
}
```

```python
df = pd.DataFrame(data_base_team)
df
```

```python
# 1) напечатать список тренеров команд, которые набирали за игру в среднем не более 50 очков и заняли место не выше 5-го
a = df.loc[(df['Среднее число очков за игру'] <= 50) & (df['Место в прошлом сезоне'] >= 5), 'Тренер']
a
```

```python
# 2) Упорядочить записи в базе в алфавитном порядке по фамилии тренера
df = df.sort_values(by='Тренер', ignore_index=True)
df
```

### База - "UFO" (14 записей)


```python
n = 14
data_base_ufo = {
    "Тип (код) UFO": np.random.randint(1, 4, size=n),
    "продолжительность наблюдения в часах": np.random.uniform(0.01, 2, size=n),
    "дата появления (день, месяц, год)": pd.to_datetime('2022-06-14') + pd.to_timedelta(
        np.random.randint(0, 366, size=n), unit="D"
    ),
    "время появления в часах": np.random.randint(0, 24, size=n),
    "Наличие световых эффектов (да, нет)": np.random.choice([True, False], size=n),
    "Наличие звуковых эффектов(да, нет)": np.random.choice([True, False], size=n)
}
```

```python
df = pd.DataFrame(data_base_ufo)
df
```

```python
# 1) напечатать список типов UFO, которые наблюдались не менее 0,1 часа и сопровождались световыми и звуковыми эффектами
a = df.loc[(df['время появления в часах'] >= 0.1) & (df['Наличие световых эффектов (да, нет)'] == True) & (df['Наличие звуковых эффектов(да, нет)'] == True), 'Тип (код) UFO']
a
```

```python
# 2) Упорядочить записи в базе по значению даты появления в порядке возрастания года, месяца, числа.
df = df.sort_values('дата появления (день, месяц, год)', ignore_index=True)
df
```
